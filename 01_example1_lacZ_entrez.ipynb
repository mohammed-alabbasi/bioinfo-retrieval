# A Single Known Gene — lacZ from NCBI


Fetch one GenBank record by its known accession (V00613, *E. coli* lacZ) with Biopython's Entrez, and save it.


```python
# Fetch the E. coli lacZ gene by its known GenBank accession, then save it
import os
from Bio import Entrez, SeqIO
from Bio.SeqRecord import SeqRecord
from Bio.SeqUtils import gc_fraction

os.makedirs("data", exist_ok=True)     # this Example creates the folders it needs 
os.makedirs("results", exist_ok=True)

Entrez.email = "you@example.com"  # NCBI requires a contact email 
handle = Entrez.efetch(db="nucleotide", id="V00613", rettype="gb", retmode="text")
record = SeqIO.read(handle, "genbank")
handle.close()

# Keep the raw GenBank record on disk
SeqIO.write(record, "data/V00613.gb", "genbank")

# Report organism, description, and the coding sequence itself
print(f"Organism: {record.annotations['organism']}")
print(f"Description: {record.description}")

for feature in record.features:
    if feature.type == "CDS":
        cds_seq = feature.extract(record.seq)
        gc = gc_fraction(cds_seq) * 100
        print(f"CDS length: {len(cds_seq)} bp")
        print(f"CDS GC content: {gc:.1f}%")

        # Save the CDS as FASTA plus a short summary
        cds_record = SeqRecord(cds_seq, id="V00613_lacZ_CDS", description="E. coli lacZ coding sequence")
        SeqIO.write(cds_record, "data/lacZ_V00613_cds.fasta", "fasta")
        with open("results/example1_lacZ_summary.txt", "w") as f:
            f.write(f"CDS length: {len(cds_seq)} bp\nCDS GC content: {gc:.1f}%\n")
        print("Saved: data/V00613.gb, data/lacZ_V00613_cds.fasta, results/example1_lacZ_summary.txt")
        break
```
