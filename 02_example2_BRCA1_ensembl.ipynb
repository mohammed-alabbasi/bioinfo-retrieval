# Example 2: A Clinically Relevant Gene — BRCA1 from Ensembl


Retrieve BRCA1 gene annotation (ENSG00000012048) from the Ensembl REST API as JSON, and save a summary.


```python
# Fetch BRCA1 gene annotation from the Ensembl REST API and save a summary
import json
import os
import requests

os.makedirs("results", exist_ok=True)  # this Example creates the folder it needs.

server = "https://rest.ensembl.org"
ext = "/lookup/id/ENSG00000012048?content-type=application/json"

response = requests.get(server + ext)
response.raise_for_status()  # stop early on a bad HTTP status.
gene = response.json()

# Report the fields that matter for a gene-level record
print(f"Gene: {gene['display_name']}")
print(f"Chromosome: {gene['seq_region_name']}")
print(f"Location: {gene['start']:,}-{gene['end']:,}")
print(f"Biotype: {gene['biotype']}")

# Save just those fields as JSON
keys = ("id", "display_name", "seq_region_name", "start", "end", "strand", "biotype")
summary = {k: gene[k] for k in keys}
with open("results/example2_BRCA1_summary.json", "w") as f:
    json.dump(summary, f, indent=2)
print("Saved: results/example2_BRCA1_summary.json")
```
