# Example 3: A Protein Record — Insulin from UniProt


Retrieve the human insulin protein record (P01308) from UniProt's REST API, and save it.


```python
# Fetch the human insulin protein record from UniProt's REST API and save it
import json
import os
import requests

os.makedirs("data", exist_ok=True)     # this Example creates the folders it needs
os.makedirs("results", exist_ok=True)

url = "https://rest.uniprot.org/uniprotkb/P01308.json"
response = requests.get(url)
response.raise_for_status()
entry = response.json()

# Navigate the nested JSON to reach the protein name and sequence
protein_name = entry["proteinDescription"]["recommendedName"]["fullName"]["value"]
organism = entry["organism"]["scientificName"]
sequence = entry["sequence"]["value"]

print(f"Protein: {protein_name}")
print(f"Organism: {organism}")
print(f"Sequence length: {len(sequence)} aa")

# Save the sequence as FASTA and a JSON summary
with open("data/P01308_insulin.fasta", "w") as f:
    f.write(f">P01308 {protein_name} [{organism}]\n{sequence}\n")
summary = {"accession": "P01308", "protein": protein_name, "organism": organism, "length_aa": len(sequence)}
with open("results/example3_insulin_summary.json", "w") as f:
    json.dump(summary, f, indent=2)
print("Saved: data/P01308_insulin.fasta, results/example3_insulin_summary.json")
```
