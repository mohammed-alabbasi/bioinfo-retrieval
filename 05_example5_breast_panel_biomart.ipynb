# Example 5: A Gene Panel Translated to Stable IDs — biomaRt (R)


Translate five breast-cancer genes from symbols to Ensembl and Entrez IDs with biomaRt, and save the table. 


```r
# Translate a breast-cancer gene panel from symbols to stable IDs with biomaRt
library(biomaRt)

dir.create("results", showWarnings = FALSE)  # this Example creates the folder it needs.

panel <- c("BRCA1", "BRCA2", "ESR1", "ERBB2", "PGR")

# Connect to Ensembl's human gene dataset
mart <- useEnsembl(
  biomart = "genes",
  dataset = "hsapiens_gene_ensembl",
  host = "https://mart.ensembl.org"  # current Ensembl mart endpoint; www.ensembl.org triggers the broken archive lookup
)

# Ask BioMart: for these symbols, return these attributes
gene_table <- getBM(
  attributes = c("hgnc_symbol", "ensembl_gene_id", "entrezgene_id",
                 "chromosome_name", "gene_biotype"),
  filters    = c("hgnc_symbol", "chromosome_name"),
  values     = list(panel, c(1:22, "X", "Y")),  # primary chromosomes only
  mart       = mart
)

gene_table <- gene_table[order(gene_table$hgnc_symbol), ]
print(gene_table, row.names = FALSE)

write.csv(gene_table, "results/example5_breast_panel_ids.csv", row.names = FALSE)
cat("Saved: results/example5_breast_panel_ids.csv\n")
```
