# Example 6: A Whole Expression Dataset — GSE2034 with GEOquery (R)


Download GEO series GSE2034 (286 breast tumours, Affymetrix HG-U133A) as an ExpressionSet and save its sample table.


```r
# Retrieve a whole GEO expression dataset (matrix + sample metadata) with GEOquery
library(GEOquery)
library(Biobase)

dir.create("data", showWarnings = FALSE)     # creates the folders it needs
dir.create("results", showWarnings = FALSE)

options(timeout = 600)  # the download is large; allow up to 10 minutes 

# Download the series matrix into data/ and load it as an ExpressionSet
gse  <- getGEO("GSE2034", GSEMatrix = TRUE, destdir = "data")
gset <- gse[[1]]

cat("Platform:", annotation(gset), "\n")
cat("Features (probes):", nrow(exprs(gset)), "\n")
cat("Samples:", ncol(exprs(gset)), "\n")

# Save the sample table (one row per patient sample)
write.csv(pData(gset), "results/example6_GSE2034_sample_metadata.csv")
cat("Saved: results/example6_GSE2034_sample_metadata.csv (raw series matrix kept in data/)\n")
```
